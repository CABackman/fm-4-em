```python
import sys
from pathlib import Path
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt

modules_path = Path.cwd().parent / "modules"
sys.path.append(str(modules_path))
from umap_plotter import *
```

# Plot UMAPS from embeddings 

## APP2024 Data

```python
emb_dir = "../02_EMBEDDINGS/APP2024/pca256"

embds = {}
# Get all embeddings corresponding to the APP2024 dataset 
for f in Path(emb_dir).glob("*.npy"):
    embds[f.stem] = np.load(f)
print(embds.keys())

# Get data
app2024 = pd.read_csv("../01_DATASETS/APP24.csv").drop("Unnamed: 0", axis=1)
smiles = app2024["SMILES"]
ys = app2024["Density [g/cc]"]
app2024.head()
```

```python
for key, item in embds.items():
    fig, ax = generate_UMAP(
        item,
        ys,
        y_prop="Crystal Density (g/cc)",
        fig_title=key,
        figsize = (8,6)
    )
    plt.savefig(f"umaps/APP2024/pca256/{key}")
    #plt.show()
```
